```python
from alfarouqai import AI, Client, environment, model_plan
print(environment())
print(model_plan(70, precision_bits=4).to_dict())
```

```python
from pathlib import Path
import tempfile
workspace = tempfile.TemporaryDirectory()
out = Path(workspace.name)
result = AI("حلل هذا النص", text="الذكاء الاصطناعي يساعد الطلاب. Python helps students.")
print(result.to_dict())
result.export(out / "text-report.json")
assert (out / "text-report.json").is_file()
```

```python
csv = out / "sales.csv"
csv.write_text("month,sales\n1,100\n2,130\n3,150\n", encoding="utf-8")
profile = Client().run("data.profile", data=csv)
print(profile.to_dict())
assert profile.data["columns"]["sales"]["mean"] > 120
```
